```python
import json
import os
import pandas as pd
```

```python
path = os.getcwd()
basePath = path.split('\\notebooks')[0]
datasetPath = os.path.join(basePath, 'Dataset')
```

```python
dataset = []
with open(os.path.join(datasetPath, 'dataset.json'), 'r') as f:
    dataset = json.load(f)

variance = []
with open(os.path.join(datasetPath, 'datasetVariance.json'), 'r') as f:
    variance = json.load(f)
```

```python
# Create a DataFrame from the dataset and variance
df = pd.DataFrame(dataset)
df_variance = pd.DataFrame(variance)

df_combined = pd.concat([df, df_variance])

# reset the index
df_combined.reset_index(drop=True, inplace=True)
```

```python
print("Dataset Information:")
df.info()
print("\nVariance Information:")
df_variance.info()
print("\nCombined Dataset Information:")
df_combined.info()
```

```python
def extract_non_ascii(text):
    """Extract non-ASCII characters from a string."""
    return ''.join(char for char in text if ord(char) > 127)

non_ascii_map = {}

for column in df_combined.columns:
    if df_combined[column].dtype == 'object':
        non_ascii_values = df_combined[column].apply(extract_non_ascii)
        non_ascii_map[column] = non_ascii_values[non_ascii_values != ''].unique().tolist()
print("\nNon-ASCII Characters Found:")
for column, non_ascii_chars in non_ascii_map.items():
    if non_ascii_chars:
        print(f"{column}: {non_ascii_chars}")
```

```python
# Define a dictionary to replace non-ASCII characters
replace_map = {
    '\u202f': ' ',  # Narrow no-break space -> Replace with normal space
    '\xa0': ' ',    # No-break space -> Replace with normal space
    '‑': '-',       # Soft hyphen -> Replace with regular hyphen
    '–': '-',       # En dash -> Replace with hyphen
    '“': '"',       # Left quote -> Replace with double quote
    '”': '"',       # Right quote -> Replace with double quote
    '‘': "'",       # Left single quote -> Replace with single quote
    '’': "'",       # Right single quote -> Replace with single quote
    '‑‑': '--',     # Double hyphen -> Replace with double hyphen
    '−': '-',       # Minus sign -> Replace with hyphen
}

df_combined_cleaned = df_combined.copy()
# Replace non-ASCII characters in the DataFrame
for column in df_combined_cleaned.columns:
    if df_combined_cleaned[column].dtype == 'object':
        df_combined_cleaned[column] = df_combined_cleaned[column].replace(replace_map, regex=True)
```

```python
df_combined_cleaned.to_csv(os.path.join(datasetPath, 'datasetWithVariance.csv'), index=False)
```
